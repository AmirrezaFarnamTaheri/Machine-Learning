$$\textbf{Supervised Learning:}$$

Supervised learning is the kind of supervision which you feed your model with labeled train data. The objective of the model would be to label the new unseen data correctly by what model has learned from train data.

Spam detecting or detecting an appropriate or harmful contents on social media are applications of supervised learning. For example in social media content moderation the model would be trained on a dataset of content which are labeled as whether they are harmful or inappropriate or not and the model will be able to detect this kind of content when facing a new data.

Other applications of supervised learning would be regression prediction. For example when you want to predict the house prices, you would feed the model with house data which also include their feature and you will label them with the target price.

$$\textbf{Unsupervised Learning:}$$

In this tyoe of learning, the computer should figure out what data wants to say by its' own. In this type of learning there is no categorized or labeled data.

Unsupervised learning is useful when we don't have ant initial information (such as groups or patterns) about data, so we just feed the model with lots of data and will wait for the model to find out the patterns by its own.

Clustering is on of the usual applications of unsupervised learning.

$$ \textbf{Self-supervised Learning}$$

We use this learning to label a fully unlabeled dataset using the data itself.
For example it could be used to repair imaages or to train LLM models.

$$\textbf{Semi-supervised Learning:}$$

Semi-supervised learning is some learning between Supervised and Unsupervised learning. It uses both labeled and unlabeled data. The model uses standard techniques to learn on labeled data, then it uses unlabelde data for generalizing and learn additional information. We use this because, labeling is costly procedure.

Some services such as Google Photos uses this supervision.



$$\textbf{Reinforcement Learning:}$$

This branch of machine learning focuses to train an agent in an environment to maximize its reward. It will be trained by interacting with environment and recieves feedback as rewared or punishment and it should try to make a strategy for interacting with the environment to maximize the rewards and minimize the punishments.

For example Alpha Go's robot is trained with reinforcement learning.

$$\textbf{Mean Absolute Error (MAE)}$$

The Mean Absolute Error (MAE) is defined as:


$$MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$$

MAE measures the average of the absolute value of error (difference between real costs and predicted cost)

Example: If the MAE is 10, it means that, the difference between predictions and actual value is 10, on average.

$$\textbf{Mean Squared Error (MSE)}$$

The Mean Squared Error (MSE) is defined as:

$$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

MSE is similar to MAE, but it squares the differences between the actual and predicted values. Which makes MSe more sensitive to outliers. It is actually penalizing the model for large errors, unlike MAE in which each errors considered equal.

MSE measures the average squared differnece between actual and predicted values and it is usually used in regression analysis

$$\textbf{Mean Absolute Percentage Error (MAPE)}$$

The Mean Absolute Percentage Error (MAPE) is defined as:

$MAPE = \frac{1}{n} \sum_{i=1}^{n} \left| \frac{y_i - \hat{y}_i}{y_i} \right| \times 100$

MAPE measures the average percentage difference between the actual and predicted values. MAPE is useful when we want to find out relative errors, for example in sale or financial forecasting.\
It is notable that, when real values are close to zero, using this error might cause some problems, since dividing by small numbers, may cause large error values.

$$\textbf{R2-Score}$$

The R2-Score, measures the proportion of the variance of dependent variables that can be explained by independent variables in our model. It is usually used to measure the performance of the regressiom model.

$$R^2 = 1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{n} (y_i - \bar{y})^2}$$

$R2$ will return us a number in the scale of 0-1, which higher number means better performance.

### Feature Scaling and Transformation

According to book:
"There are two common ways to get all attributes to have the same scale: min-max scaling (normalization) and standardization."

<font color='#D61E85' size='+2'>**Q15:**</font> Why we need feature scaling at all? Between these two common ways which one is more robust to outlier? What happened if we don’t do this in terms of convergence of the model?

<font color='#8FCF26' size='+2'>**A15:**</font>

Machine learning algorithms don't usually perform well when the input numerical attributes have very different scales.

Without any scaling, most models will be biased towards the data with larger scales.

Min-Max scaling and Standardization are two common ways of feature scaling.

Min-max scaling (aka normalization) is the simplest: for
each attribute, the values are shifted and rescaled so that they end up ranging
from 0 to 1.  This is performed by subtracting the min value and dividing by
the difference between the min and the max.

Standardization is different: first it subtracts the mean value (so standardized
values have a zero mean), then it divides the result by the standard deviation
(so standardized values have a standard deviation equal to 1). Unlike min-max scaling, standardization does not restrict values to a specific range.
However, standardization is much less affected by outliers. 

If we don't use featuer scaling, the model might converge very slower or don't converge at all.

*Transformers* definition:

> Some estimators (such as a `SimpleImputer`) can also transform a dataset; these are called *transformers*. Once again, the API is simple: the transformation is performed by the `transform()` method with the dataset to transform as a parameter. It returns the transformed dataset. This transformation generally relies on the learned parameters, as is the case for a `SimpleImputer`. All transformers also have a convenience method called `fit_transform()`, which is equivalent to calling `fit()` and then `transform()` (but sometimes `fit_transform()` is optimized and runs much faster).

The formula for Robust Scaling is:

$$X_{scaled} = \dfrac{X - median(X)}{IQR(X)}$$

Where:

$\textbf{X}$ is the input feature vector.\
$\textbf{median(X)}$ is the median of the feature vector.\
$\textbf{IQR(X)}$ is the IQR: The interquartile range of the data $(Q3 - Q1)$


```python
from sklearn.base import BaseEstimator, TransformerMixin

class RobustScaler(BaseEstimator, TransformerMixin):
    """RobustScaler is a custom class for performing robust scaling on input data.

    Args:
        BaseEstimator (type): Base class for all estimators in scikit-learn.
        TransformerMixin (type): Mixin class for all transformers in scikit-learn.
    """
    def __init__(self):
        """Initialize the RobustScaler object."""
        self.feature_names_in_ = None
        self.n_features_in_ = None
        self.median_ = None
        self.iqr_ = None
    
    def fit(self, X, y=None):
        """Fit the RobustScaler to the input data.

        Args:
            X (array-like or DataFrame): Input data.
            y (array-like, optional): Target data. Defaults to None.

        Returns:
            self: Fitted RobustScaler object.
        """
        if isinstance(X, pd.DataFrame):
            self.feature_names_in_ = np.array(X.columns)
        self.n_features_in_ = X.shape[1]
        self.median_ = np.median(X, axis=0)
        self.iqr_ = np.percentile(X, 75, axis=0) - np.percentile(X, 25, axis=0)
        return self
    
    def transform(self, X):
        """Transform the input data using robust scaling.

        Args:
            X (array-like or DataFrame): Input data.

        Returns:
            array-like: Transformed data.
        """
        return (X - self.median_) / self.iqr_
    
    def inverse_transform(self, X):
        """Reverse the robust scaling operation on the input data.

        Args:
            X (array-like or DataFrame): Scaled data.

        Returns:
            array-like: Inverse transformed data.
        """
        return (X * self.iqr_) + self.median_
    
    def get_feature_names_out(self, input_features=None):
        """Get the feature names of the transformed data.

        Args:
            input_features (array-like, optional): Input feature names. Defaults to None.

        Raises:
            ValueError: If the length of input_features does not match the number of input features.
            ValueError: If input_features does not match the feature_names_in_ attribute.

        Returns:
            array-like: Transformed feature names.
        """
        if input_features is not None:
            if len(input_features) != self.n_features_in_:
                raise ValueError("Length of input_features does not match n_features_in_.")
            if self.feature_names_in_ is not None and not np.array_equal(input_features, self.feature_names_in_):
                raise ValueError("input_features does not match feature_names_in_.")
            return input_features
        else:
            if self.feature_names_in_ is not None:
                return self.feature_names_in_
            else:
                return np.array([f"x{i}" for i in range(self.n_features_in_)])


# ChatGPT is used for dubugging, improving and also writing documentation for this class
```

### Randomized Search

```python
from sklearn.model_selection import RandomizedSearchCV
import random as random
```

$$\textbf{Randomized Search Linear Regression Model}$$

```python
full_pipeline = Pipeline([
    ("linear_regression", LinearRegression())
])

param_distribs = {
    'linear_regression__copy_X': [True, False],
    'linear_regression__fit_intercept': [True, False],
    'linear_regression__n_jobs': np.random.randint(1, 100, size=1000),
    'linear_regression__normalize': [True, False],
    'linear_regression__positive': [True, False]
}

rnd_search_lr = RandomizedSearchCV(
    full_pipeline, param_distributions=param_distribs, n_iter=250, cv=10,
    scoring='r2', random_state=42
)

rnd_search_lr.fit(train_features, train_labels)

best_rnd_lr_model = rnd_search_lr.best_estimator_
```

```python
# Get the results and sort them by the R^2 score in descending order
results = pd.DataFrame(rnd_search_lr.cv_results_)
results = results.sort_values(by='mean_test_score', ascending=False)

# Select the top 40 models
top_40_models = results.head(40)

# Create a DataFrame to display the features
df_top_40_models = pd.DataFrame(top_40_models['params'])
df_top_40_models.insert(0, 'Rank', range(1, len(df_top_40_models) + 1))

# Extract the features from the dictionary and create separate columns
df_top_40_models = pd.concat([df_top_40_models.drop(['params'], axis=1), df_top_40_models['params'].apply(pd.Series)], axis=1)

# Drop 'linear_regression__' from column names
df_top_40_models.columns = df_top_40_models.columns.str.replace('linear_regression__', '')

# Add 'test_score' column
df_top_40_models['test_score'] = top_40_models['mean_test_score']

# Convert 'n_jobs' column to numeric type
df_top_40_models['n_jobs'] = pd.to_numeric(df_top_40_models['n_jobs'], errors='coerce')

# Round 'n_jobs', 'test_score', and other columns to 3 decimal places
df_top_40_models['n_jobs'] = df_top_40_models['n_jobs'].round(0)
df_top_40_models['test_score'] = df_top_40_models['test_score'].round(3)

print(df_top_40_models)
```

```python
predicted_labels_rnd_lr = exp_transformer.transform(best_rnd_lr_model.predict(test_features))
rnd_search_lr_mae = mean_absolute_error(test_labels, predicted_labels_rnd_lr)
rnd_search_lr_mse = mean_squared_error(test_labels, predicted_labels_rnd_lr)
rnd_search_lr_mape = mean_absolute_percentage_error(test_labels, predicted_labels_rnd_lr)
rnd_search_lr_r2 = r2_score(test_labels, predicted_labels_rnd_lr)

print("Random Search LR - MAE:", rnd_search_lr_mae)
print("Random Search LR - MSE:", rnd_search_lr_mse)
print("Random Search LR - MAPE:", rnd_search_lr_mape)
print("Random Search LR - R^2:", rnd_search_lr_r2)
```
